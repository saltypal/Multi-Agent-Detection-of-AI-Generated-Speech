```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

DRIVE_DIR = Path('/content/drive/MyDrive/40_PER_22_Data')
CSV_PATH = DRIVE_DIR / 'spectral_features.csv'

print(f"Loading data from {CSV_PATH}...")
df = pd.read_csv(CSV_PATH)
print(f"Total shape: {df.shape}")
```

```python
# Download spectral_model.py to use training logic
!wget -q https://raw.githubusercontent.com/saltypal/Multi-Agent-Detection-of-AI-Generated-Speech/main/spectral/spectral_model.py
```

```python
from spectral_model import train_model

train_df = df[df['split'] == 'train'].copy()
test_df = df[df['split'] == 'test'].copy()

print(f"Train size: {len(train_df)}")
print(f"Test size: {len(test_df)}")

OUT_DIR = DRIVE_DIR / 'spectral_results'
OUT_DIR.mkdir(exist_ok=True)

model = train_model(
    train_df=train_df, 
    test_df=test_df, 
    out_dir=OUT_DIR, 
    use_gpu=True # Set to False if not running on GPU runtime
)
```
