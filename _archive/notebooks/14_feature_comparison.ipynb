## Concept and Math

High correlation indicates potential redundancy.
Complementary feature groups are valuable for fusion and robustness.


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Replace synthetic blocks with real outputs from earlier notebooks.
rng = np.random.default_rng(42)
X_spec = rng.normal(size=(200, 20))
X_pros = rng.normal(size=(200, 8))
X_ling = rng.normal(size=(200, 12))
X = np.concatenate([X_spec, X_pros, X_ling], axis=1)

df = pd.DataFrame(X)
mean_abs_corr = df.corr().abs().values[np.triu_indices(df.shape[1], k=1)].mean()
print("mean_abs_corr:", mean_abs_corr)

Xs = StandardScaler().fit_transform(X)
proj = PCA(n_components=2).fit_transform(Xs)
print("PCA projection shape:", proj.shape)
```
