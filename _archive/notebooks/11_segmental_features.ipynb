## Concept and Math

Formants are resonant peaks of the vocal tract transfer function.
F1 and F2 encode major vowel-space structure and speaker articulation traits.


```python
from pathlib import Path
import numpy as np
import parselmouth

DATA_ROOT = Path("../dataset")
audio_files = sorted(DATA_ROOT.rglob("*.flac")) + sorted(DATA_ROOT.rglob("*.wav"))
if not audio_files:
    raise FileNotFoundError("No audio found under ../dataset")

snd = parselmouth.Sound(str(audio_files[0]))
frm = snd.to_formant_burg(time_step=0.01)

times = np.arange(0.05, max(0.06, snd.duration - 0.05), 0.01)
f1 = np.array([frm.get_value_at_time(1, t) for t in times], dtype=float)
f2 = np.array([frm.get_value_at_time(2, t) for t in times], dtype=float)
f3 = np.array([frm.get_value_at_time(3, t) for t in times], dtype=float)

print("F1 mean:", np.nanmean(f1))
print("F2 mean:", np.nanmean(f2))
print("F3 mean:", np.nanmean(f3))
```
