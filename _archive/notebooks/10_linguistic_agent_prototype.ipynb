## Concept and Math

Linguistic cues combine what was said with how it was temporally aligned.
Mismatches in timing regularity and lexical behavior may indicate synthesis.


```python
from pathlib import Path
import numpy as np
from transformers import pipeline

DATA_ROOT = Path("../dataset")
audio_files = sorted(DATA_ROOT.rglob("*.flac")) + sorted(DATA_ROOT.rglob("*.wav"))
if not audio_files:
    raise FileNotFoundError("No audio found under ../dataset")

asr = pipeline("automatic-speech-recognition", model="openai/whisper-tiny", return_timestamps="word")
out = asr(str(audio_files[0]))
text = out.get("text", "")
chunks = out.get("chunks", [])

durations = []
for c in chunks:
    ts = c.get("timestamp")
    if ts and ts[0] is not None and ts[1] is not None:
        durations.append(ts[1] - ts[0])

print("text:", text)
print("word_count:", len(chunks))
print("mean_word_duration:", np.mean(durations) if durations else np.nan)
```
