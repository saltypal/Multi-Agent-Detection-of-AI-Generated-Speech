Let's now design a Robust loader for any sample size, for any channels 

```python
def load_audio(path):
    audio,sr = lb.load(path, sr=None, mono=False)

    if audio.ndim==1:
        waveform = torch.from_numpy(audio).unsqueeze(0) # then my shape will come (1, Samples)
    
    else:
        waveform = torch.from_numpy(audio)
    
    return waveform, sr
```
