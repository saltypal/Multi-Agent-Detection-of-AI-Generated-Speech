## Concept and Math

Let each agent output a fake score in [0, 1].
Agreement can be approximated by low cross-agent variance per sample.
Disagreement flags uncertain or out-of-distribution behavior.


```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(0)
S = pd.DataFrame({
    "spectral": np.clip(rng.normal(0.55, 0.18, 300), 0, 1),
    "prosodic": np.clip(rng.normal(0.50, 0.20, 300), 0, 1),
    "linguistic": np.clip(rng.normal(0.52, 0.22, 300), 0, 1),
})
S["mean_score"] = S.mean(axis=1)
S["agreement_std"] = S[["spectral", "prosodic", "linguistic"]].std(axis=1)

print(S.describe())
print("high_disagreement:", (S["agreement_std"] > 0.25).sum())
```
