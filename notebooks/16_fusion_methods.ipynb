## Concept and Math

Average fusion: s = mean_i s_i.
Weighted fusion: s = sum_i w_i s_i with sum_i w_i = 1.
Voting fusion: each agent votes after thresholding; majority decides.


```python
import numpy as np

rng = np.random.default_rng(123)
S = np.clip(rng.normal(0.5, 0.2, size=(200, 3)), 0, 1)
y = rng.integers(0, 2, size=200)

def fuse_avg(scores):
    return scores.mean(axis=1)

def fuse_weighted(scores, w):
    w = np.asarray(w, dtype=float)
    w = w / (w.sum() + 1e-8)
    return scores @ w

def fuse_vote(scores, thr=0.5):
    return ((scores >= thr).sum(axis=1) >= 2).astype(int)

p_avg = (fuse_avg(S) >= 0.5).astype(int)
p_w = (fuse_weighted(S, [0.5, 0.3, 0.2]) >= 0.5).astype(int)
p_v = fuse_vote(S)

print("acc_avg:", (p_avg == y).mean())
print("acc_weighted:", (p_w == y).mean())
print("acc_vote:", (p_v == y).mean())
```

## PyTorch Equivalent Snippet
Understand the librosa block first, then map it to this snippet.


```python
import torch

scores = torch.rand(200, 3)
w = torch.tensor([0.5, 0.3, 0.2])
w = w / w.sum()
fused = scores @ w
pred = (fused >= 0.5).int()
print(fused.shape, pred.shape)
```
