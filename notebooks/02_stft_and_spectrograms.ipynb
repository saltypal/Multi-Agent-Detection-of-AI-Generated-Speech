# Notebook 02 - STFT and Spectrograms

## Goal
Convert waveform into a time-frequency representation using STFT.


## Agenda
- Define frame and hop in samples
- Compute STFT
- Convert to dB scale
- Visualize spectrogram


## Concept and Math

STFT applies FFT on short windows: X(m, k) = sum_n x[n] w[n-mH] e^{-j2pi kn/N}.
Speech is non-stationary globally but approximately stationary in short frames.


```python
from pathlib import Path
import numpy as np
import librosa as lb
import librosa.display
import matplotlib.pyplot as plt

DATA_ROOT = Path("../dataset")
audio_files = sorted(DATA_ROOT.rglob("*.flac")) + sorted(DATA_ROOT.rglob("*.wav"))
if not audio_files:
    raise FileNotFoundError("No .flac or .wav found under ../dataset")

audio_path = audio_files[0]
print(f"Using: {audio_path}")

wave, sr = lb.load(audio_path, sr=16000, mono=True)
win = int(0.025 * sr)
hop = int(0.010 * sr)

stft = lb.stft(wave, n_fft=512, win_length=win, hop_length=hop)
mag_db = lb.amplitude_to_db(np.abs(stft), ref=np.max)

plt.figure(figsize=(10, 4))
librosa.display.specshow(mag_db, sr=sr, hop_length=hop, x_axis="time", y_axis="hz", cmap="magma")
plt.colorbar(format="%+2.0f dB")
plt.title("STFT Magnitude Spectrogram")
plt.tight_layout()
plt.show()
```

## PyTorch Equivalent Snippet
Understand the librosa block first, then map it to this snippet.


```python
import torch
import torchaudio

wave_t, sr_t = torchaudio.load(str(audio_path))
wave_t = wave_t.mean(dim=0)
stft_t = torch.stft(
    wave_t,
    n_fft=512,
    hop_length=int(0.010 * sr_t),
    win_length=int(0.025 * sr_t),
    window=torch.hann_window(int(0.025 * sr_t)),
    return_complex=True,
)
print(stft_t.shape)
```
