## Concept and Math

DFT formula: X[k] = sum_{n=0}^{N-1} x[n] exp(-j2pi kn/N).
FFT computes the same values faster using divide-and-conquer structure.
Bin frequency is f_k = k f_s / N.


```python
import numpy as np
import matplotlib.pyplot as plt

fs, N = 64, 64
n = np.arange(N)
x = np.sin(2 * np.pi * 8 * n / fs)

k = n.reshape((N, 1))
W = np.exp(-2j * np.pi * k * n / N)
X_dft = W @ x
X_fft = np.fft.fft(x)

print("max error:", np.max(np.abs(X_dft - X_fft)))
print("peak bin:", np.argmax(np.abs(X_fft[:N//2])))

plt.stem(np.abs(X_fft[:N//2]), basefmt=" ")
plt.title("FFT Magnitude (Positive Bins)")
plt.xlabel("Bin")
plt.ylabel("Magnitude")
plt.show()
```

## PyTorch Equivalent Snippet
Understand the librosa block first, then map it to this snippet.


```python
import torch

fs, N = 64, 64
n = torch.arange(N)
x = torch.sin(2 * torch.pi * 8 * n / fs)
X = torch.fft.fft(x)
print(torch.argmax(torch.abs(X[:N//2])).item())
```
